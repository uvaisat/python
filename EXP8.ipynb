```python
#write a python program to accept 5 student's names and roll numbers, and map these into a dictionary(roll as key, name as value).
#Sort the dictonary in ascending and desending order of their roll number[use sorted()]
students = {}
roll1 = int(input("Enter roll number 1: "))
name1 = input("Enter name 1: ")
students[roll1] = name1
roll2 = int(input("Enter roll number 2: "))
name2 = input("Enter name 2: ")
students[roll2] = name2
roll3 = int(input("Enter roll number 3: "))
name3 = input("Enter name 3: ")
students[roll3] = name3
roll4 = int(input("Enter roll number 4: "))
name4 = input("Enter name 4: ")
students[roll4] = name4
roll5 = int(input("Enter roll number 5: "))
name5 = input("Enter name 5: ")
students[roll5] = name5
print("Original dictionary:", students)
asc = dict(sorted(students.items()))
print("Ascending order:", asc)
des = dict(sorted(students.items(), reverse=True))
print("Descending order:", des)
```

```python
#write a python program to accept 2 dictionaries(dic1 contains 3 student's names and roll numbers, and dic2 contains 2 student's names and roll numbers).
#Merge 2 dictionaries and display it[use update()]
data1=input("Enter 3 students roll number and names(appropriate order seperated by space:)").split()
dic1={
    data1[0]:data1[1],
    data1[2]:data1[3],
    data1[4]:data1[5]
}
data2=input("Enter 2 students roll number and names(appropriate order seperated by space:").split()
dic2={
    data2[0]:data2[1],
    data2[2]:data2[3]
}
print("Dictionary 1:",dic1)
print("Dictionary 2:",dic2)
dic1.update(dic2)
print("Merged dictionary:",dic1)
```
