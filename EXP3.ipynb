```python
#Write a python program to change a given string to a new string where the first and last  chars have been exchanged[eg: python->nythop]
x=str(input("Enter a string :"))
print("New string :",x[-1]+x[1:-1]+x[0])
```

```python
#Write a Python program to get a string made of the first 2 and the last 2 chars from a given a string
x=str(input("Enter a string :"))
print("New string:",x[:2]+x[-2:])
```

```python
#write a python program to create a single string seperated with space from two strings by swapping the character at position 1
#[eg: s1=hello, s2=world, after swap= wello horld]
x=str(input("Enter first string :"))
y=str(input("Enter second string :"))
a=y[0]+x[1:]
b=x[0]+y[1:]
print("New String:",a+" "+b)
```

```python
#write a python program to get a string from a given string where all occurrences of its first char have been changed to $ except the first char itself
#(use replace())[eg: onion->oni$n]
x=str(input("Enter a string :"))
a=x[0]
b=x[1:].replace(a,"$")
print("Replaced result :",a+b)
```
