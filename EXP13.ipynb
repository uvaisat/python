```python
%%writefile Rectangle.py
def area(length, breadth):
    return length * breadth
def perimeter(length, breadth):
    return 2 * (length + breadth)
```

```python
%%writefile Circle.py
import math
def area(radius):
    return math.pi * radius * radius
def perimeter(radius):
    return 2 * math.pi * radius
```

```python
# Importing the rectangle module
import Rectangle
import Circle
while True:
    print("1.Rectangle")
    print("2.Circle")
    print("3.Exit")
    ch=int(input("Enter your choice:"))
    if ch==1:
        l = float(input("Enter length: "))
        b = float(input("Enter breadth: "))
        print("Rectangle Area:",round(Rectangle.area(l,b),2))
        print("Rectangle Perimeter:",round(Rectangle.perimeter(l,b),2))
        print()
    elif ch==2:
        r = float(input("Enter radius: "))
        print("Circle  Area:",round(Circle.area(r),2))
        print("Circle  perimeter:",round(Circle.perimeter(r),2))
        print()
    elif ch ==3:
        print("Exiting program...")
        break
    else:
        print("Enter valid Choice")
        print()
```
