```python
#Create a list of colors from comma-seperated  color names entered by user.
#Display first and last colors
a=input("Enter a list of colors seperated by comma:").split(",")
print(a)
print("First color:",a[0].strip())
print("Last color:",a[-1].strip())
```

```python
#Write a python program to accept a filename from the user, print extension of that.
file=input("Enter the file name: ")
ext=file.split(".")
print("Extention :",ext[-1])
```

```python
#Write a python program which accepts a sequence of comma-seperated numbers from the user.
#Generate a list with those numbers.
#Find the sum of all items in that list.
num=input("Enter a list of numbers seperated by comma :").split(",")
num=[int(x) for x in num]
print("List :",num)
print("Sum of numbers in the list :",sum(num))
```
