```python
#Write a Python program using regular expressions to extract all email addresses from a given text.
import re
with open("emails.txt", "r") as file:
    text = file.read()
emails = re.findall(r'[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Za-z]{2,}', text)
print("Email addresses:")
print(emails)
```
