```python
%%writefile player.txt
Messi
Ronaldo
Neymar
Suarez
```

```python
#Read a file line by line and store it in a list.
lines = []
with open("player.txt","r") as file:
    for line in file:
        lines.append(line.strip())
print("List:",lines)
```

```python
%%writefile input.txt
Lion
Tiger
Elephant
Giraffe
Zebra
Deer
```

```python
#Copy Odd Lines to a New File
with open("input.txt","r") as src:
    lines=src.readlines()
with open("odd_lines.txt","w") as dest:
    for i in range(len(lines)):
        if i % 2 ==0:
            dest.write(lines[i])
print("Odd lines copied successfully!")
```

```python
#Read CSV File
import csv
with open("student.csv","r") as file:
    reader=csv.reader(file)
    for row in reader:
        print(row)
```

```python
#Read specific columns in a CSV file
import csv
f1name=input("Enter the CSV filename:")
with open(f1name,"r")as f:
    reader=csv.DictReader(f)
    print("Student Details:")
    for row in reader:
        print("Name:",row['Name'],"\tCourse:",row['Course'])
```

```python
# Write a python dictionary to CSV file,then read and display its contents
import csv
studentdetail=[
    {'Name':'Shahan','Age':21,'Course':'MCA'},
    {'Name':'Arjun','Age':20,'Course':'BCA'},
    {'Name':'Arun','Age':20,'Course':'BSc CS'}]
with open("studentdetail.csv","w",newline="") as file:
    fields=['Name','Age','Place']
    writer=csv.DictWriter(file,fieldnames=fields)
    writer.writeheader()
    writer.writerows(studentdetail)
print("Dictionary written successfully!")
```
