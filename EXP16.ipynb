```python
#Write a Python program to demonstrate exception handling by raising a ValueError when the user enters an invalid input.
try:
    name = input("Enter Name: ")
    if not name.replace(" ", "").isalpha():
        raise ValueError("Invalid name! Enter alphabets only.")
    age = int(input("Enter Age: "))
    if age < 0:
        raise ValueError("Invalid age! Age cannot be negative.")
    mark = float(input("Enter Mark: "))
    if mark < 0:
        raise ValueError("Invalid Mark! Mark cannot be negative.")
    if mark > 100:
        raise ValueError("Invalid Mark! Mark cannot be over 100.")
    print("\nName:", name)
    print("Age:", age)
    print("Mark :", mark)
except ValueError as e:
    print("Error:", e)
finally:
    print("Program execution completed.")
```
