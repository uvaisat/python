```python
#Write a python program to get the 3rd element and 3rd element from the last of a tuple
num= tuple(input("Enter a list of numbers seperated by comma :").split(","))
print("Tuple :",num)
print("3rd element from first in the tuple :",num[2])
print("3rd element from last in the tuple :",num[-3])
```

```python
#Write a pyton program which accepts a sequence of comma-seperated numbers from user
#Generate a tuple with those numbers
#Find average of all items in that tuple
num=tuple(map(int,input("Enter a list of numbers seperated by comma :").split(",")))
print("Tuple :",num)
average= sum(num)/len(num)
print("Average of numbers in the tuple :",round(average,1))
```
