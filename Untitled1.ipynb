```python
#input the radius of a circle and compute the area
import math
r=int (input("Enter the radius:"))
area=math.pi *r*r
print("Area of the circle",round(area,2))
```

```python
#Write a python program that accepts an integer(n) and computes the value of n+nn+nnn.
n=input("Enter an integer")
nn=n+n
nnn=n+n+n
sum=int(n)+int(nn)+int(nnn)
print("Result :",n,"+",nn,"+",nnn,"=",sum)
```

```python
#Write a program to find the biggest of 3 numbers(Use max())
a=int(input("Enter the first Number:"))
b=int(input("Enter the second Number:"))
c=int(input("Enter the third Number:"))
x=max(a,b,c)
print("The biggest number is",x)
```
