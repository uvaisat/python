```python
#List comprehensions:
#a- Generate +ve list of numbers from a given list of integers
numbers = list(map(int,input("Enter list of numbers seperated by space:").split()))
positive = [n for n in numbers if n > 0]
print("Positive numbers:", positive)
```

```python
#b-Square of N numbers
n = int(input("Enter total numbers: "))
numbers = list(map(int,input(f"Enter {n} numbers seperated by space:").split()))
squares = [i**2 for i in numbers]
print("Square of",n,"numbers:", squares)
```

```python
#c-Form a list of vowels selected from a given word
word = input("Enter a word: ")
vowels = [i for i in word if i.lower() in "aeiou"]
print("Vowels:",vowels)
```

```python
#d-List ordinal value of each element of a word(Hint:Use ord() to get ordinal values)
word = input("Enter a word: ")
ordinal= [ord(ch) for ch in word]
print("Ordinal values:", ordinal)
```
