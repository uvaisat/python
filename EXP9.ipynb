```python
#Write a python program to add "ing" to the end of a given string.
#If the string already ends with "ing",add "ly" instead
x=input("Enter a string:")
if x[-3:] == "ing":
    print("New string: ",x+"ly")
else:
    print("New string: ",x+"ing")
```

```python
#Write a Python program to check the given string is palindrome or not.
x=input("Enter a string:")
if x== x[::-1]:
    print("Result: The given string is a Palindrome")
else:
    print("Result: The given string is not a Palindrome")
```
